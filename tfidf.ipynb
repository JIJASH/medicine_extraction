```python
# 🎯 COMPLETE STEP-BY-STEP EXPLANATION: Why All Documents Are Essential
print("🔍 UNDERSTANDING THE COMPLETE TF-IDF PROCESS")
print("=" * 80)

# Our example data
ocr_input = "Paracetamo1"
database_medicines = ["Paracetamol", "Ibuprofen", "Aspirin", "Amoxicillin", "Diclofenac"]
all_documents = [ocr_input] + database_medicines

print("\n🎯 THE FUNDAMENTAL QUESTION:")
print("   Why can't we just compare 'Paracetamo1' directly with 'Paracetamol'?")
print("   Why do we need Ibuprofen, Aspirin, etc.?")

print("\n📚 STEP 1: UNDERSTANDING THE VECTOR SPACE CONCEPT")
print("-" * 50)
print("   TF-IDF creates a 'Vector Space' where:")
print("   • Each unique n-gram becomes a DIMENSION")
print("   • Each document becomes a VECTOR in this space")
print("   • We need ALL documents to define the complete space!")

# Create n-grams for all documents
def create_ngrams(text, n_min=2, n_max=4):
    text = text.lower()
    ngrams = []
    for n in range(n_min, n_max + 1):
        for i in range(len(text) - n + 1):
            ngram = text[i:i+n]
            if any(c.isalpha() for c in ngram):
                ngrams.append(ngram)
    return ngrams

all_ngrams_per_doc = [create_ngrams(doc) for doc in all_documents]

# Get ALL unique n-grams across ALL documents
all_unique_ngrams = set()
for ngrams in all_ngrams_per_doc:
    all_unique_ngrams.update(ngrams)

print(f"\n🔢 VECTOR SPACE DIMENSIONS:")
print(f"   Total unique n-grams across ALL documents: {len(all_unique_ngrams)}")
print("   Each n-gram = 1 dimension in our vector space")

# Show what happens if we only use 2 documents vs all documents
only_two_docs = [ocr_input, "Paracetamol"]
two_docs_ngrams = [create_ngrams(doc) for doc in only_two_docs]
unique_ngrams_two_docs = set()
for ngrams in two_docs_ngrams:
    unique_ngrams_two_docs.update(ngrams)

print(f"\n⚠️  COMPARISON: What if we ONLY used 2 documents?")
print(f"   Documents: ['Paracetamo1', 'Paracetamol']")
print(f"   Unique n-grams: {len(unique_ngrams_two_docs)}")
print(f"   Vector space dimensions: {len(unique_ngrams_two_docs)}")

print(f"\n✅ USING ALL 6 DOCUMENTS:")
print(f"   Documents: {len(all_documents)} total")
print(f"   Unique n-grams: {len(all_unique_ngrams)}")
print(f"   Vector space dimensions: {len(all_unique_ngrams)}")

print(f"\n📊 IMPACT: {len(all_unique_ngrams) - len(unique_ngrams_two_docs)} additional dimensions!")
print("   This creates a much richer, more discriminative vector space!")
```

```python
# 🔍 STEP 2: WHY IDF NEEDS ALL DOCUMENTS - THE CRITICAL INSIGHT
print("\n" + "=" * 80)
print("🔍 STEP 2: THE IDF CALCULATION - Why Every Document Matters")
print("=" * 80)

import math

# Function to calculate IDF for a given document collection
def calculate_idf(documents_ngrams, target_ngram):
    total_docs = len(documents_ngrams)
    docs_containing_term = sum(1 for ngrams in documents_ngrams if target_ngram in ngrams)
    
    if docs_containing_term == 0:
        return 0
    return math.log(total_docs / docs_containing_term)

# Let's see how IDF values change with different document collections
target_ngram = 'ara'  # Our example n-gram

print(f"\n🎯 ANALYZING IDF FOR N-GRAM '{target_ngram}':")

# Scenario 1: Only 2 documents (OCR + target match)
docs_2 = [ocr_input, "Paracetamol"]
ngrams_2 = [create_ngrams(doc) for doc in docs_2]
idf_2_docs = calculate_idf(ngrams_2, target_ngram)

print(f"\n📊 SCENARIO 1: Only 2 documents ['Paracetamo1', 'Paracetamol']")
print(f"   Documents containing '{target_ngram}': 2 out of 2")
print(f"   IDF('{target_ngram}') = log(2/2) = log(1) = {idf_2_docs:.4f}")

# Scenario 2: All 6 documents
docs_6 = all_documents
ngrams_6 = all_ngrams_per_doc
idf_6_docs = calculate_idf(ngrams_6, target_ngram)

print(f"\n📊 SCENARIO 2: All 6 documents {[doc[:8]+'...' if len(doc)>8 else doc for doc in docs_6]}")
print(f"   Documents containing '{target_ngram}': 2 out of 6")
print(f"   IDF('{target_ngram}') = log(6/2) = log(3) = {idf_6_docs:.4f}")

print(f"\n💡 CRITICAL INSIGHT:")
print(f"   With 2 documents: IDF = {idf_2_docs:.4f} (very low - term seems common)")
print(f"   With 6 documents: IDF = {idf_6_docs:.4f} (higher - term is actually rare!)")
print(f"   Difference: {idf_6_docs - idf_2_docs:.4f}")

print(f"\n🎯 WHY THIS MATTERS:")
print("   • With only 2 docs: 'ara' appears in 100% of documents → looks common → low importance")
print("   • With all 6 docs: 'ara' appears in 33% of documents → actually rare → high importance")
print("   • The additional documents help us understand what's truly rare vs common!")

# Let's see another example with a different n-gram
other_ngram = 'ibu'  # Only in Ibuprofen
idf_ibu_2 = calculate_idf(ngrams_2, other_ngram)
idf_ibu_6 = calculate_idf(ngrams_6, other_ngram)

print(f"\n🔬 ANOTHER EXAMPLE: N-gram '{other_ngram}' (only in 'Ibuprofen')")
print(f"   With 2 docs: IDF('{other_ngram}') = {idf_ibu_2:.4f} (doesn't appear in our 2-doc set)")
print(f"   With 6 docs: IDF('{other_ngram}') = {idf_ibu_6:.4f} (appears in 1/6 documents)")
print("   Without all documents, we'd miss important discriminative information!")
```

```python
# 🚀 STEP 3: THE COMPLETE VECTOR CREATION PROCESS
print("\n" + "=" * 80)
print("🚀 STEP 3: HOW VECTORS ARE CREATED AND COMPARED")
print("=" * 80)

from collections import Counter

# Functions matching your system exactly
def compute_term_frequency(ngrams):
    total_terms = len(ngrams)
    if total_terms == 0:
        return {}
    term_counts = Counter(ngrams)
    tf = {}
    for term, count in term_counts.items():
        tf[term] = count / total_terms
    return tf

def compute_inverse_document_frequency(documents_ngrams):
    total_docs = len(documents_ngrams)
    if total_docs == 0:
        return {}
    
    all_terms = set()
    for ngrams in documents_ngrams:
        all_terms.update(ngrams)
    
    idf = {}
    for term in all_terms:
        docs_containing_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
        if docs_containing_term > 0:
            idf[term] = math.log(total_docs / docs_containing_term)
        else:
            idf[term] = 0
    return idf

def compute_tfidf_vector(tf, idf, all_terms):
    tfidf_vector = []
    for term in all_terms:
        tf_value = tf.get(term, 0)
        idf_value = idf.get(term, 0)
        tfidf_value = tf_value * idf_value
        tfidf_vector.append(tfidf_value)
    return tfidf_vector

print("🔧 BUILDING THE COMPLETE VECTOR SPACE:")

# Step 1: Calculate IDF using ALL documents
print("\n   Step 1: Calculate IDF using ALL 6 documents")
idf_values = compute_inverse_document_frequency(all_ngrams_per_doc)
all_terms = sorted(list(idf_values.keys()))
print(f"   ✅ IDF calculated for {len(idf_values)} unique terms")

# Step 2: Build TF-IDF vectors for each document
print("\n   Step 2: Build TF-IDF vectors for each document")
tf_values = []
tfidf_vectors = []

for i, doc_ngrams in enumerate(all_ngrams_per_doc):
    # Calculate TF for this document
    tf = compute_term_frequency(doc_ngrams)
    tf_values.append(tf)
    
    # Calculate TF-IDF vector
    tfidf_vector = compute_tfidf_vector(tf, idf_values, all_terms)
    tfidf_vectors.append(tfidf_vector)
    
    non_zero_count = sum(1 for val in tfidf_vector if val > 0)
    doc_name = all_documents[i][:15] + "..." if len(all_documents[i]) > 15 else all_documents[i]
    print(f"   Document {i} ({doc_name}): {non_zero_count}/{len(all_terms)} non-zero dimensions")

print(f"\n✅ RESULT: {len(tfidf_vectors)} vectors in {len(all_terms)}-dimensional space")

# Step 3: Compare OCR input with each database medicine
print("\n   Step 3: Compare OCR input vector with each database medicine vector")
ocr_vector = tfidf_vectors[0]  # OCR input vector

def compute_cosine_similarity(vector1, vector2):
    if len(vector1) != len(vector2):
        return 0
    
    dot_product = sum(a * b for a, b in zip(vector1, vector2))
    magnitude1 = math.sqrt(sum(a * a for a in vector1))
    magnitude2 = math.sqrt(sum(a * a for a in vector2))
    
    if magnitude1 == 0 or magnitude2 == 0:
        return 0
    
    return dot_product / (magnitude1 * magnitude2)

print(f"\n📊 SIMILARITY RESULTS:")
for i in range(1, len(tfidf_vectors)):  # Skip OCR input (index 0)
    db_vector = tfidf_vectors[i]
    similarity = compute_cosine_similarity(ocr_vector, db_vector)
    medicine_name = all_documents[i]
    
    # Highlight best match
    if similarity == max([compute_cosine_similarity(ocr_vector, tfidf_vectors[j]) for j in range(1, len(tfidf_vectors))]):
        status = "🎯 BEST MATCH!"
    else:
        status = ""
    
    print(f"   '{ocr_input}' vs '{medicine_name}': {similarity:.6f} ({similarity*100:.2f}%) {status}")

print(f"\n🎯 THE MAGIC HAPPENS:")
print("   • All documents define the vector space dimensions")
print("   • IDF weights are calculated from the complete collection")
print("   • Each medicine gets a unique position in this space")
print("   • Similar medicines end up close together")
print("   • Cosine similarity measures the angle between vectors")
```

## 🔢 How TF-IDF Vectors are Used in Cosine Similarity

**Perfect Question!** Now let's see exactly how those TF-IDF values become vectors and how cosine similarity uses them for comparison.

We'll trace through the complete process with your exact example data.

```python
# 🚀 STEP-BY-STEP: From TF-IDF Values to Cosine Similarity
print("🔢 COMPLETE WALKTHROUGH: TF-IDF Vectors → Cosine Similarity")
print("=" * 80)

# Your example data
ocr_input = "Paracetamo1"
target_medicine = "Paracetamol"

# Let's work with a simplified document collection for clarity
simplified_docs = [ocr_input, target_medicine, "Ibuprofen"]
print(f"📚 SIMPLIFIED EXAMPLE (3 documents):")
for i, doc in enumerate(simplified_docs):
    print(f"   Document {i}: '{doc}'")

# Step 1: Create n-grams for each document
def create_ngrams(text, n_min=2, n_max=4):
    text = text.lower()
    ngrams = []
    for n in range(n_min, n_max + 1):
        for i in range(len(text) - n + 1):
            ngram = text[i:i+n]
            if any(c.isalpha() for c in ngram):
                ngrams.append(ngram)
    return ngrams

all_doc_ngrams = [create_ngrams(doc) for doc in simplified_docs]

print(f"\n🔤 N-GRAM EXTRACTION:")
for i, (doc, ngrams) in enumerate(zip(simplified_docs, all_doc_ngrams)):
    print(f"   Doc {i} '{doc}': {len(ngrams)} n-grams")
    print(f"      First 10: {ngrams[:10]}")

# Step 2: Get ALL unique n-grams (vector dimensions)
all_unique_ngrams = set()
for ngrams in all_doc_ngrams:
    all_unique_ngrams.update(ngrams)

all_unique_ngrams = sorted(list(all_unique_ngrams))  # Sort for consistent ordering
print(f"\n📐 VECTOR SPACE DIMENSIONS:")
print(f"   Total unique n-grams: {len(all_unique_ngrams)}")
print(f"   Each n-gram becomes one dimension in our vector space")
print(f"   Sample dimensions: {all_unique_ngrams[:15]}...")

print(f"\n💡 KEY INSIGHT:")
print(f"   Every document will be represented as a {len(all_unique_ngrams)}-dimensional vector")
print(f"   Each dimension corresponds to one unique n-gram")
```

```python
# 🔢 STEP 3: Calculate TF-IDF Values and Build Vectors
print("\n" + "=" * 80)
print("🔢 STEP 3: TF-IDF CALCULATION AND VECTOR CONSTRUCTION")
print("=" * 80)

import math
from collections import Counter

# Calculate TF for each document
def compute_tf(ngrams):
    total_terms = len(ngrams)
    term_counts = Counter(ngrams)
    tf = {}
    for term, count in term_counts.items():
        tf[term] = count / total_terms
    return tf

# Calculate IDF for all terms
def compute_idf(documents_ngrams):
    total_docs = len(documents_ngrams)
    all_terms = set()
    for ngrams in documents_ngrams:
        all_terms.update(ngrams)
    
    idf = {}
    for term in all_terms:
        docs_containing_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
        idf[term] = math.log(total_docs / docs_containing_term) if docs_containing_term > 0 else 0
    return idf

# Calculate all TF values
print("📊 CALCULATING TF (Term Frequency) for each document:")
tf_values = []
for i, ngrams in enumerate(all_doc_ngrams):
    tf = compute_tf(ngrams)
    tf_values.append(tf)
    print(f"   Document {i} ('{simplified_docs[i]}'): {len(tf)} unique terms")

# Calculate IDF values
print(f"\n📈 CALCULATING IDF (Inverse Document Frequency):")
idf_values = compute_idf(all_doc_ngrams)
print(f"   IDF calculated for {len(idf_values)} unique terms")

# Show specific example calculations
target_ngram = 'ara'
if target_ngram in idf_values:
    docs_with_ara = sum(1 for ngrams in all_doc_ngrams if target_ngram in ngrams)
    print(f"\n🎯 DETAILED CALCULATION FOR '{target_ngram}':")
    print(f"   Documents containing '{target_ngram}': {docs_with_ara}/{len(simplified_docs)}")
    print(f"   IDF('{target_ngram}') = log({len(simplified_docs)}/{docs_with_ara}) = {idf_values[target_ngram]:.4f}")
    
    # Show TF for each document
    for i, tf in enumerate(tf_values):
        tf_val = tf.get(target_ngram, 0)
        tfidf_val = tf_val * idf_values[target_ngram]
        print(f"   Doc {i}: TF='{tf_val:.4f}', TF-IDF={tfidf_val:.4f}")

# Build TF-IDF vectors
print(f"\n⚡ BUILDING TF-IDF VECTORS:")
print(f"   Each vector has {len(all_unique_ngrams)} dimensions")

tfidf_vectors = []
for i, tf in enumerate(tf_values):
    vector = []
    for term in all_unique_ngrams:  # This ensures consistent ordering
        tf_val = tf.get(term, 0)  # 0 if term not in this document
        idf_val = idf_values.get(term, 0)
        tfidf_val = tf_val * idf_val
        vector.append(tfidf_val)
    
    tfidf_vectors.append(vector)
    non_zero_count = sum(1 for val in vector if val > 0)
    print(f"   Vector {i} ('{simplified_docs[i]}'): {non_zero_count}/{len(vector)} non-zero values")

print(f"\n📋 VECTOR SUMMARY:")
print(f"   • Vector 0 ('{ocr_input}'): OCR input vector")
print(f"   • Vector 1 ('{target_medicine}'): Target medicine vector") 
print(f"   • Vector 2 ('Ibuprofen'): Different medicine vector")
print(f"   • Each vector: {len(all_unique_ngrams)} dimensions")
```

```python
# 📏 STEP 4: COSINE SIMILARITY CALCULATION - The Final Step!
print("\n" + "=" * 80)
print("📏 STEP 4: COSINE SIMILARITY - Vector Comparison")
print("=" * 80)

def compute_cosine_similarity_detailed(vector1, vector2, show_steps=True):
    """Cosine similarity with detailed step-by-step calculation"""
    
    if show_steps:
        print(f"   Vector 1 length: {len(vector1)}")
        print(f"   Vector 2 length: {len(vector2)}")
    
    # Step 1: Dot Product (A · B)
    dot_product = sum(a * b for a, b in zip(vector1, vector2))
    
    # Show some non-zero contributions to dot product
    if show_steps:
        non_zero_products = [(i, a, b, a*b) for i, (a, b) in enumerate(zip(vector1, vector2)) if a*b != 0]
        print(f"\n   🔢 DOT PRODUCT CALCULATION (A · B):")
        print(f"      Total terms contributing: {len(non_zero_products)}")
        if non_zero_products:
            print(f"      Sample contributions:")
            for i, (idx, a, b, product) in enumerate(non_zero_products[:5]):  # Show first 5
                ngram = all_unique_ngrams[idx]
                print(f"        '{ngram}': {a:.6f} × {b:.6f} = {product:.6f}")
            if len(non_zero_products) > 5:
                print(f"        ... and {len(non_zero_products)-5} more")
        
        print(f"      Final dot product: {dot_product:.8f}")
    
    # Step 2: Magnitude of vector1 (||A||)
    magnitude1 = math.sqrt(sum(a * a for a in vector1))
    
    # Step 3: Magnitude of vector2 (||B||)
    magnitude2 = math.sqrt(sum(b * b for b in vector2))
    
    if show_steps:
        print(f"\n   📏 MAGNITUDE CALCULATIONS:")
        print(f"      ||Vector1|| = √(Σa²) = {magnitude1:.8f}")
        print(f"      ||Vector2|| = √(Σb²) = {magnitude2:.8f}")
    
    # Step 4: Cosine similarity
    if magnitude1 == 0 or magnitude2 == 0:
        cosine_sim = 0
    else:
        cosine_sim = dot_product / (magnitude1 * magnitude2)
    
    if show_steps:
        print(f"\n   🎯 FINAL COSINE SIMILARITY:")
        print(f"      cos(θ) = (A·B) / (||A|| × ||B||)")
        print(f"      cos(θ) = {dot_product:.8f} / ({magnitude1:.8f} × {magnitude2:.8f})")
        print(f"      cos(θ) = {dot_product:.8f} / {magnitude1 * magnitude2:.8f}")
        print(f"      cos(θ) = {cosine_sim:.8f}")
        print(f"      Similarity: {cosine_sim*100:.4f}%")
    
    return cosine_sim

# Compare OCR input with target medicine
ocr_vector = tfidf_vectors[0]      # "Paracetamo1"
target_vector = tfidf_vectors[1]   # "Paracetamol"
different_vector = tfidf_vectors[2] # "Ibuprofen"

print(f"🎯 COMPARISON 1: OCR '{ocr_input}' vs Target '{target_medicine}'")
print("-" * 60)
similarity_target = compute_cosine_similarity_detailed(ocr_vector, target_vector, show_steps=True)

print(f"\n🎯 COMPARISON 2: OCR '{ocr_input}' vs Different 'Ibuprofen'")
print("-" * 60)
similarity_different = compute_cosine_similarity_detailed(ocr_vector, different_vector, show_steps=True)

print(f"\n📊 FINAL RESULTS SUMMARY:")
print("=" * 50)
print(f"   '{ocr_input}' vs '{target_medicine}':  {similarity_target:.6f} ({similarity_target*100:.2f}%)")
print(f"   '{ocr_input}' vs 'Ibuprofen':         {similarity_different:.6f} ({similarity_different*100:.2f}%)")

if similarity_target > similarity_different:
    print(f"\n✅ CORRECT MATCH IDENTIFIED!")
    print(f"   The system correctly identifies '{target_medicine}' as the best match")
    print(f"   Despite the OCR error 'l'→'1', similarity is still high!")
else:
    print(f"\n❌ Unexpected result - check the calculation")
```

## 3. Step 1: Character N-gram Creation

**Your system uses character n-grams (2-4 characters) instead of words**. This is brilliant for OCR errors!

Let's see how n-grams are created from our example:

```python
def create_ngrams(text, n_min=2, n_max=4):
    """Create character n-grams exactly like in your system"""
    text = text.lower()  # Convert to lowercase
    ngrams = []
    
    for n in range(n_min, n_max + 1):  # n = 2, 3, 4
        for i in range(len(text) - n + 1):
            ngram = text[i:i+n]
            if any(c.isalpha() for c in ngram):  # Only if contains letters
                ngrams.append(ngram)
    
    return ngrams

# Create n-grams for our example
ocr_ngrams = create_ngrams(ocr_extracted)
paracetamol_ngrams = create_ngrams("Paracetamol")

print(f"📝 N-GRAMS FOR '{ocr_extracted}' (OCR input):")
print(f"   Total n-grams: {len(ocr_ngrams)}")
print(f"   2-grams: {[ng for ng in ocr_ngrams if len(ng) == 2]}")
print(f"   3-grams: {[ng for ng in ocr_ngrams if len(ng) == 3]}")
print(f"   4-grams: {[ng for ng in ocr_ngrams if len(ng) == 4]}")

print(f"\n📝 N-GRAMS FOR 'Paracetamol' (Database):")
print(f"   Total n-grams: {len(paracetamol_ngrams)}")
print(f"   2-grams: {[ng for ng in paracetamol_ngrams if len(ng) == 2]}")
print(f"   3-grams: {[ng for ng in paracetamol_ngrams if len(ng) == 3]}")
print(f"   4-grams: {[ng for ng in paracetamol_ngrams if len(ng) == 4]}")

# Show overlap (this is why similarity works!)
common_ngrams = set(ocr_ngrams) & set(paracetamol_ngrams)
print(f"\n🎯 COMMON N-GRAMS (why they match): {len(common_ngrams)}/{len(set(ocr_ngrams))}")
print(f"   Common: {sorted(common_ngrams)}")
print(f"   Overlap percentage: {len(common_ngrams)/len(set(ocr_ngrams))*100:.1f}%")
```

## 4. Step 2: Create Complete Document Collection

## 5. Step 3: Calculate Term Frequency (TF)

**Mathematical Formula**: `TF(term, document) = count(term) / total_terms_in_document`

Let's calculate TF for our OCR input and the target medicine:

```python
def compute_term_frequency(ngrams):
    """Calculate TF exactly like your system"""
    total_terms = len(ngrams)
    if total_terms == 0:
        return {}
    
    term_counts = Counter(ngrams)
    tf = {}
    
    # TF = (Number of times term appears) / (Total number of terms)
    for term, count in term_counts.items():
        tf[term] = count / total_terms
    
    return tf

# Calculate TF for OCR input (Document 0)
ocr_tf = compute_term_frequency(documents_ngrams[0])

# Calculate TF for Paracetamol (Document 1)
paracetamol_tf = compute_term_frequency(documents_ngrams[1])

print("📊 TERM FREQUENCY (TF) CALCULATION:")
print("\n🔍 OCR Input 'Paracetamo1':")
print(f"   Total n-grams: {len(documents_ngrams[0])}")
print("   Top 10 TF values:")
for i, (term, tf_val) in enumerate(sorted(ocr_tf.items(), key=lambda x: x[1], reverse=True)[:10]):
    print(f"     '{term}': {tf_val:.4f} (appears {int(tf_val * len(documents_ngrams[0]))} times)")

print("\n💊 Database 'Paracetamol':")
print(f"   Total n-grams: {len(documents_ngrams[1])}")
print("   Top 10 TF values:")
for i, (term, tf_val) in enumerate(sorted(paracetamol_tf.items(), key=lambda x: x[1], reverse=True)[:10]):
    print(f"     '{term}': {tf_val:.4f} (appears {int(tf_val * len(documents_ngrams[1]))} times)")

# Show some common terms and their TF values
common_terms = set(ocr_tf.keys()) & set(paracetamol_tf.keys())
print(f"\n🎯 COMMON TERMS TF COMPARISON ({len(common_terms)} common terms):")
for term in sorted(list(common_terms))[:8]:  # Show first 8
    print(f"   '{term}': OCR={ocr_tf[term]:.4f}, DB={paracetamol_tf[term]:.4f}")
```

## 6. Step 4: Calculate Inverse Document Frequency (IDF)

**Mathematical Formula**: `IDF(term) = log(total_documents / documents_containing_term)`

IDF gives higher weights to rare terms and lower weights to common terms.

```python
def compute_inverse_document_frequency(documents_ngrams):
    """Calculate IDF exactly like your system"""
    total_docs = len(documents_ngrams)
    if total_docs == 0:
        return {}
    
    # Get all unique terms across all documents
    all_terms = set()
    for ngrams in documents_ngrams:
        all_terms.update(ngrams)
    
    idf = {}
    
    # IDF = log(Total documents / Documents containing term)
    for term in all_terms:
        docs_containing_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
        if docs_containing_term > 0:
            idf[term] = math.log(total_docs / docs_containing_term)
        else:
            idf[term] = 0
    
    return idf

# Calculate IDF for all terms
idf = compute_inverse_document_frequency(documents_ngrams)

print(f"📈 INVERSE DOCUMENT FREQUENCY (IDF) CALCULATION:")
print(f"   Total unique terms across all documents: {len(idf)}")
print(f"   Total documents: {len(documents_ngrams)}")
print()

# Show IDF values with interpretation
print("🔍 IDF VALUES (higher = more rare/important):")
sorted_idf = sorted(idf.items(), key=lambda x: x[1], reverse=True)

print("   HIGHEST IDF (rarest terms):")
for i, (term, idf_val) in enumerate(sorted_idf[:8]):
    docs_with_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
    print(f"     '{term}': {idf_val:.4f} (appears in {docs_with_term}/{len(documents_ngrams)} documents)")

print("\n   LOWEST IDF (most common terms):")
for i, (term, idf_val) in enumerate(sorted_idf[-8:]):
    docs_with_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
    print(f"     '{term}': {idf_val:.4f} (appears in {docs_with_term}/{len(documents_ngrams)} documents)")

# Show IDF for some key terms from our example
key_terms = ['par', 'ara', 'rac', 'ace', 'cet', 'eta', 'tam', 'amo']
print("\n🎯 IDF VALUES FOR KEY PARACETAMOL N-GRAMS:")
for term in key_terms:
    if term in idf:
        docs_with_term = sum(1 for ngrams in documents_ngrams if term in ngrams)
        print(f"   '{term}': {idf[term]:.4f} (in {docs_with_term} docs)")
```

## 7. Step 5: Compute TF-IDF Vectors

**Mathematical Formula**: `TF-IDF(term, document) = TF(term, document) × IDF(term)`

Now we combine TF and IDF to create the final vectors!

```python
def compute_tfidf_vector(tf, idf, all_terms):
    """Compute TF-IDF vector exactly like your system"""
    tfidf_vector = []
    
    # TF-IDF = TF * IDF
    for term in all_terms:
        tf_value = tf.get(term, 0)
        idf_value = idf.get(term, 0)
        tfidf_value = tf_value * idf_value
        tfidf_vector.append(tfidf_value)
    
    return tfidf_vector

# Get all unique terms (this defines our vector space dimensions)
all_terms = sorted(list(idf.keys()))

print(f"⚡ TF-IDF VECTOR COMPUTATION:")
print(f"   Vector space dimensions: {len(all_terms)} (each unique n-gram is one dimension)")
print()

# Compute TF for all documents first
all_tf = []
for i, doc_ngrams in enumerate(documents_ngrams):
    tf = compute_term_frequency(doc_ngrams)
    all_tf.append(tf)

# Compute TF-IDF vectors for all documents
tfidf_vectors = []
for i, tf in enumerate(all_tf):
    tfidf_vector = compute_tfidf_vector(tf, idf, all_terms)
    tfidf_vectors.append(tfidf_vector)
    non_zero = sum(1 for val in tfidf_vector if val > 0)
    doc_name = all_documents[i]
    print(f"   Document {i} ('{doc_name}'): {non_zero}/{len(all_terms)} non-zero values")

# Let's examine the TF-IDF values for our key comparison
ocr_vector = tfidf_vectors[0]      # OCR input vector
paracetamol_vector = tfidf_vectors[1]  # Paracetamol database vector

print("\n🎯 TOP TF-IDF VALUES COMPARISON:")
print("\n   OCR 'Paracetamo1' - Top 10 TF-IDF values:")
ocr_tfidf_terms = [(all_terms[i], ocr_vector[i]) for i in range(len(all_terms)) if ocr_vector[i] > 0]
ocr_tfidf_terms.sort(key=lambda x: x[1], reverse=True)
for term, tfidf_val in ocr_tfidf_terms[:10]:
    print(f"     '{term}': {tfidf_val:.6f}")

print("\n   Database 'Paracetamol' - Top 10 TF-IDF values:")
para_tfidf_terms = [(all_terms[i], paracetamol_vector[i]) for i in range(len(all_terms)) if paracetamol_vector[i] > 0]
para_tfidf_terms.sort(key=lambda x: x[1], reverse=True)
for term, tfidf_val in para_tfidf_terms[:10]:
    print(f"     '{term}': {tfidf_val:.6f}")

print(f"\n✅ TF-IDF vectors computed! Ready for similarity calculation.")
```

## 8. Step 6: Calculate Cosine Similarity

**Mathematical Formula**: `cosine_similarity = (A · B) / (||A|| × ||B||)`

Where:
- A · B = dot product = Σ(A[i] × B[i])
- ||A|| = magnitude = √(Σ(A[i]²))
- ||B|| = magnitude = √(Σ(B[i]²))

Let's calculate this step by step!

```python
def compute_cosine_similarity(vector1, vector2, show_calculation=False):
    """Compute cosine similarity exactly like your system with detailed steps"""
    if len(vector1) != len(vector2):
        return 0
    
    # Step 1: Dot product
    dot_product = sum(a * b for a, b in zip(vector1, vector2))
    
    # Step 2: Magnitude of vector1
    magnitude1 = math.sqrt(sum(a * a for a in vector1))
    
    # Step 3: Magnitude of vector2  
    magnitude2 = math.sqrt(sum(a * a for a in vector2))
    
    if show_calculation:
        print(f"   📐 Dot product (A·B): {dot_product:.8f}")
        print(f"   📏 Magnitude of A (||A||): {magnitude1:.8f}")
        print(f"   📏 Magnitude of B (||B||): {magnitude2:.8f}")
    
    # Step 4: Avoid division by zero
    if magnitude1 == 0 or magnitude2 == 0:
        return 0
    
    # Step 5: Cosine similarity = dot_product / (magnitude1 * magnitude2)
    cosine_sim = dot_product / (magnitude1 * magnitude2)
    
    if show_calculation:
        print(f"   🎯 Cosine similarity: {dot_product:.8f} / ({magnitude1:.8f} × {magnitude2:.8f}) = {cosine_sim:.8f}")
    
    return cosine_sim

print("📏 COSINE SIMILARITY CALCULATION:")
print()

# Calculate similarity between OCR input and each database medicine
print("🎯 DETAILED CALCULATION: OCR 'Paracetamo1' vs Database 'Paracetamol':")
similarity = compute_cosine_similarity(ocr_vector, paracetamol_vector, show_calculation=True)
print(f"   ✅ FINAL SIMILARITY SCORE: {similarity:.6f} ({similarity*100:.2f}%)")

print("\n📊 SIMILARITY SCORES WITH ALL DATABASE MEDICINES:")
similarities = []
for i in range(1, len(tfidf_vectors)):  # Skip index 0 (OCR input)
    sim_score = compute_cosine_similarity(ocr_vector, tfidf_vectors[i])
    similarities.append(sim_score)
    medicine_name = all_documents[i]
    status = "🎯 BEST MATCH" if sim_score == max(similarities) else ""
    print(f"   {medicine_name:12} : {sim_score:.6f} ({sim_score*100:.2f}%) {status}")

# Find the best match
best_match_idx = similarities.index(max(similarities))
best_match = database_medicines[best_match_idx]
best_score = max(similarities)

print(f"\n🏆 FINAL RESULT:")
print(f"   OCR Input: '{ocr_extracted}'")
print(f"   Best Match: '{best_match}' with {best_score*100:.2f}% similarity")
print(f"   Match Quality: {'EXCELLENT' if best_score > 0.8 else 'GOOD' if best_score > 0.6 else 'MODERATE' if best_score > 0.4 else 'POOR'}")
```

## 9. Step 7: Understanding Why This Works for OCR Errors

Let's analyze **why** character n-grams make your system robust to OCR errors:

```python
# Detailed analysis of why OCR error matching works
def analyze_ngram_overlap(text1, text2):
    """Analyze n-gram overlap between two texts"""
    ngrams1 = set(create_ngrams(text1))
    ngrams2 = set(create_ngrams(text2))
    
    common = ngrams1 & ngrams2
    union = ngrams1 | ngrams2
    only_text1 = ngrams1 - ngrams2
    only_text2 = ngrams2 - ngrams1
    
    return {
        'text1_ngrams': ngrams1,
        'text2_ngrams': ngrams2,
        'common': common,
        'union': union,
        'only_text1': only_text1,
        'only_text2': only_text2,
        'jaccard_similarity': len(common) / len(union) if union else 0,
        'overlap_percentage': len(common) / min(len(ngrams1), len(ngrams2)) * 100 if ngrams1 and ngrams2 else 0
    }

print("🔍 WHY CHARACTER N-GRAMS HANDLE OCR ERRORS:")
print()

# Analyze our main example
analysis = analyze_ngram_overlap(ocr_extracted, "Paracetamol")

print(f"📊 N-GRAM OVERLAP ANALYSIS:")
print(f"   OCR '{ocr_extracted}': {len(analysis['text1_ngrams'])} unique n-grams")
print(f"   DB 'Paracetamol': {len(analysis['text2_ngrams'])} unique n-grams")
print(f"   Common n-grams: {len(analysis['common'])} ({analysis['overlap_percentage']:.1f}% overlap)")
print(f"   Jaccard similarity: {analysis['jaccard_similarity']:.3f}")

print(f"\n✅ SHARED N-GRAMS (why similarity is high):")
for ngram in sorted(analysis['common']):
    print(f"   '{ngram}'")

print(f"\n❌ OCR-ONLY N-GRAMS (caused by OCR error):")
for ngram in sorted(analysis['only_text1']):
    print(f"   '{ngram}' (contains the '1' from OCR error)")

print(f"\n❌ DATABASE-ONLY N-GRAMS (correct spelling):")
for ngram in sorted(analysis['only_text2']):
    print(f"   '{ngram}' (contains the correct 'l')")

print("\n🎯 KEY INSIGHT:")
print(f"   Even with OCR error 'l'→'1', we still have {analysis['overlap_percentage']:.1f}% n-gram overlap!")
print("   This high overlap creates strong TF-IDF similarity, enabling correct matching.")

# Compare with a completely different medicine
print("\n🔬 COMPARISON WITH DIFFERENT MEDICINE:")
analysis_diff = analyze_ngram_overlap(ocr_extracted, "Ibuprofen")
print(f"   '{ocr_extracted}' vs 'Ibuprofen': {analysis_diff['overlap_percentage']:.1f}% overlap")
print(f"   This low overlap results in low TF-IDF similarity, correctly rejecting the match.")
```

## 10. Complete Medicine Matching Pipeline

Let's implement the complete pipeline exactly as in your system:

```python
def custom_tfidf_cosine_similarity(search_term, document_list):
    """Complete TF-IDF pipeline exactly like your system"""
    print("🔧 Starting custom TF-IDF and Cosine Similarity computation...")
    
    # Step 1: Prepare all documents (search term + document list)
    all_documents = [search_term] + document_list
    print(f"📄 Processing {len(all_documents)} documents (1 query + {len(document_list)} targets)")
    
    # Step 2: Create n-grams for all documents
    print("🔤 Creating character n-grams...")
    documents_ngrams = []
    for i, doc in enumerate(all_documents):
        ngrams = create_ngrams(doc, n_min=2, n_max=4)
        documents_ngrams.append(ngrams)
    
    # Step 3: Compute IDF for all terms
    print("📊 Computing Inverse Document Frequency (IDF)...")
    idf = compute_inverse_document_frequency(documents_ngrams)
    print(f"   Total unique terms: {len(idf)}")
    
    # Step 4: Get all unique terms for vector space
    all_terms = list(idf.keys())
    print(f"🔢 Vector space dimension: {len(all_terms)}")
    
    # Step 5: Compute TF-IDF vectors for all documents
    print("⚡ Computing TF-IDF vectors...")
    tfidf_vectors = []
    for i, ngrams in enumerate(documents_ngrams):
        tf = compute_term_frequency(ngrams)
        tfidf_vector = compute_tfidf_vector(tf, idf, all_terms)
        tfidf_vectors.append(tfidf_vector)
    
    # Step 6: Compute cosine similarities
    print("🎯 Computing cosine similarities...")
    query_vector = tfidf_vectors[0]  # First vector is the search term
    similarities = []
    
    for i in range(1, len(tfidf_vectors)):  # Skip the query vector
        similarity = compute_cosine_similarity(query_vector, tfidf_vectors[i])
        similarities.append(similarity)
    
    print("✅ Custom TF-IDF and Cosine Similarity computation completed!")
    return similarities

# Test the complete pipeline with various OCR errors
test_cases = [
    ("Paracetamo1", "l→1 error"),       # Our main example
    ("PARACETAMOL", "Case variation"),   # Case difference
    ("Paracetaml", "Missing letter"),   # Missing 'o'
    ("Paracetarnol", "OCR rn→m error"), # 'm' read as 'rn'
    ("lbuprofen", "I→l error"),         # Different medicine with error
    ("Asprin", "Missing letter"),       # Different medicine, missing 'i'
]

print("🧪 TESTING COMPLETE PIPELINE WITH VARIOUS OCR ERRORS:")
print("=" * 80)

for ocr_input, error_type in test_cases:
    print(f"\n🔍 Test Case: '{ocr_input}' ({error_type})")
    
    # Run the complete pipeline
    similarities = custom_tfidf_cosine_similarity(ocr_input.lower(), 
                                                   [med.lower() for med in database_medicines])
    
    # Find best match
    best_idx = similarities.index(max(similarities))
    best_match = database_medicines[best_idx]
    best_score = max(similarities)
    
    print(f"   🎯 Best Match: '{best_match}' ({best_score*100:.2f}% similarity)")
    
    # Show top 3 matches
    sorted_results = sorted(zip(database_medicines, similarities), 
                          key=lambda x: x[1], reverse=True)
    print("   📊 Top 3 matches:")
    for i, (med, score) in enumerate(sorted_results[:3]):
        print(f"      {i+1}. {med}: {score*100:.2f}%")
    print("-" * 60)
```
